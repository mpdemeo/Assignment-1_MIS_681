```python
import random
 
# Game 1: Rock Paper Scissors
def rock_paper_scissors():
    options = ["rock", "paper", "scissors"]
    player = input("Choose rock, paper, or scissors: ").lower().strip()
 
    if player not in options:
        print("Invalid choice. Please try again.")
        return
 
    computer = random.choice(options)
    print("Computer chose:", computer)
 
    if player == computer:
        print("It's a tie!")
    elif (player == "rock" and computer == "scissors") or \
         (player == "paper" and computer == "rock") or \
         (player == "scissors" and computer == "paper"):
        print("You win!")
    else:
        print("Computer wins!")
 
 
# Game 2: Draw a shape using a symbol
def draw_shape():
    print("1. Rectangle")
    print("2. Triangle")
    shape = input("Choose a shape (1 or 2): ").strip()
    symbol = input("Enter a symbol (such as * or #): ").strip()
 
    if not symbol:
        print("Please enter a symbol.")
        return
 
    try:
        rows = int(input("Enter the number of rows: "))
        if rows <= 0:
            print("Rows must be positive.")
            return
 
        if shape == "1":
            columns = int(input("Enter the number of columns: "))
            if columns <= 0:
                print("Columns must be positive.")
                return
            for i in range(rows):
                print(symbol * columns)
        elif shape == "2":
            for i in range(1, rows + 1):
                print(symbol * i)
        else:
            print("Invalid shape choice.")
    except ValueError:
        print("Please enter whole numbers for dimensions.")
 
 
# Game 3: Break a dollar amount into bills and coins
def currency_calculator():
    try:
        amount = float(input("Enter a dollar amount: $"))
        if amount < 0:
            print("Please enter a positive amount.")
            return
 
        cents = round(amount * 100)
        dollars = cents // 100
        cents = cents % 100
        quarters = cents // 25
        cents = cents % 25
        dimes = cents // 10
        cents = cents % 10
        nickels = cents // 5
        pennies = cents % 5
 
        print("Dollars:", dollars)
        print("Quarters:", quarters)
        print("Dimes:", dimes)
        print("Nickels:", nickels)
        print("Pennies:", pennies)
    except ValueError:
        print("Please enter a valid amount, such as 2.75.")
 
 
# Main menu: keep showing it until the user chooses Exit
def main():
    input("Begin the Application (press Enter): ")
 
    while True:
        print("\nMenu Options:")
        print("1. Rock Paper Scissors")
        print("2. Draw a Shape")
        print("3. Currency Calculator")
        print("4. Exit")
 
        choice = input("Enter a choice: ").strip()
        print()
 
        if choice == "1":
            rock_paper_scissors()
        elif choice == "2":
            draw_shape()
        elif choice == "3":
            currency_calculator()
        elif choice == "4":
            print("Good Bye!!!")
            break
        else:
            print("Invalid choice. Enter 1, 2, 3, or 4.")
 
 
# Start the program
main()
```
